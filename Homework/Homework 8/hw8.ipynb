### Problem 1(d) — Estimating a Linear Model from the Factorial Design

Assume the response is approximated locally by the linear model

$$
r = \beta_0 + \beta_x x + \beta_y y + \beta_z z
$$

where $x,y,z \in \{-1,+1\}$.

---

Using a full $2^3$ factorial design with coded variables $\{-1,+1\}$, the intercept is the overall average response:

$$
\beta_0 = \bar r
$$

and each main-effect coefficient is half of the corresponding factorial effect:

$$
\beta_x = \frac{\text{Effect}_x}{2}, \qquad
\beta_y = \frac{\text{Effect}_y}{2}, \qquad
\beta_z = \frac{\text{Effect}_z}{2}
$$

From part (c),

$$
\text{Effect}_x = 5.9350
$$

$$
\text{Effect}_y = 5.6840
$$

$$
\text{Effect}_z = 0.7563
$$

So,

$$
\beta_x = \frac{5.9350}{2} = 2.9675
$$

$$
\beta_y = \frac{5.6840}{2} = 2.8420
$$

$$
\beta_z = \frac{0.7563}{2} = 0.3782
$$

---

Now compute the intercept from the mean of all 8 responses:

$$
\beta_0
=
\frac{6.1576 + 2.3134 + 7.6612 + 12.1116 + 11.3405 + 8.9345 + 13.4421 + 18.2671}{8}
$$

$$
\beta_0
=
\frac{80.2279}{8}
=
10.0285
$$

---

Therefore, the estimated linear model is

$$
\boxed{
r = 10.0285 + 2.9675x + 2.8420y + 0.3782z
}
$$

# Problem 2

**a.** OLS assumes linear contributions of each parameter, independent errors, homoscedasticity, and Gaussian residuals. Given the assumptions, the probability model is:

$$
\hat{y} = \beta_0 + \beta_1 x + \beta_2 y + \beta_3 z + \epsilon
$$

Ordinary least squares minimizes: 
$$ \Sigma (y_i - \hat{y_i})^2 $$
Maximizing likelihood minimizes the squared error, which maximizes the closeness to the mean, assuming Gaussian noise. <br>

**b.**
Bayesian framework means
$$Posterior \space \alpha \space (Likelihood \times Prior)$$
In a Bayesian framework, using $\beta_i ~ N(0,10)$ as the priors, with $\sigma ~ HalfNormal(5)$. Then, following whatever the sampling technique is from the dataset, we end up developing our posterior.

**c.**
weakly informative normal prior: 

$$
\beta_j \sim \mathcal{N}(0, 10)
$$

---

strongly regularizing normal prior

$$
\beta_j \sim \mathcal{N}(0, 1)
$$

---

heavy-tailed Student-$t$ prior

$$
\beta_j \sim t_{\nu=3}(0, 5)
$$

````python
# Problem 2(d) — Bayesian Linear Regression Implementation
# Three prior models: weak normal, strong normal, and Student-t

import pandas as pd
import pymc as pm
import arviz as az
import matplotlib.pyplot as plt

# --------------------------------------------------
# Load dataset
# --------------------------------------------------
data = pd.read_csv("mse238_doe_bayes_dataset.csv")

# Predictors and response using the actual CSV headers
x1 = data["x"].values
x2 = data["y"].values
x3 = data["z"].values
response = data["outcome"].values

print(data.columns.tolist())

# --------------------------------------------------
# Helper function to build and sample a Bayesian model
# --------------------------------------------------
def fit_bayesian_model(prior_type="weak_normal"):
  with pm.Model() as model:

    # Priors for regression coefficients
    if prior_type == "weak_normal":
      beta0 = pm.Normal("beta0", mu=0, sigma=10)
      beta1 = pm.Normal("beta1", mu=0, sigma=10)
      beta2 = pm.Normal("beta2", mu=0, sigma=10)
      beta3 = pm.Normal("beta3", mu=0, sigma=10)

    elif prior_type == "strong_normal":
      beta0 = pm.Normal("beta0", mu=0, sigma=1)
      beta1 = pm.Normal("beta1", mu=0, sigma=1)
      beta2 = pm.Normal("beta2", mu=0, sigma=1)
      beta3 = pm.Normal("beta3", mu=0, sigma=1)

    elif prior_type == "student_t":
      beta0 = pm.StudentT("beta0", nu=3, mu=0, sigma=5)
      beta1 = pm.StudentT("beta1", nu=3, mu=0, sigma=5)
      beta2 = pm.StudentT("beta2", nu=3, mu=0, sigma=5)
      beta3 = pm.StudentT("beta3", nu=3, mu=0, sigma=5)

    else:
      raise ValueError("Unknown prior_type")

    # Prior for residual standard deviation
    sigma = pm.HalfNormal("sigma", sigma=5)

    # Linear predictor
    mu = beta0 + beta1 * x1 + beta2 * x2 + beta3 * x3

    # Likelihood
    y_obs = pm.Normal("y_obs", mu=mu, sigma=sigma, observed=response)

    # Posterior sampling
    trace = pm.sample(
      draws=2000,
      tune=1000,
      chains=2,
      target_accept=0.9,
      return_inferencedata=True
    )

  return model, trace

# --------------------------------------------------
# Fit all three prior models
# --------------------------------------------------
models = {}
traces = {}

for prior_name in ["weak_normal", "strong_normal", "student_t"]:
  print(f"\nFitting model with prior: {prior_name}")
  model, trace = fit_bayesian_model(prior_type=prior_name)
  models[prior_name] = model
  traces[prior_name] = trace

# --------------------------------------------------
# Posterior summaries
# --------------------------------------------------
for prior_name, trace in traces.items():
  print(f"\nPosterior summary for {prior_name}:")
  print(az.summary(trace, var_names=["beta0", "beta1", "beta2", "beta3", "sigma"]))

# --------------------------------------------------
# Trace plots
# --------------------------------------------------
for prior_name, trace in traces.items():
  az.plot_trace(trace, var_names=["beta0", "beta1", "beta2", "beta3", "sigma"])
  plt.suptitle(f"Trace plots: {prior_name}", y=1.02)
  plt.show()

# --------------------------------------------------
# Posterior plots
# --------------------------------------------------
for prior_name, trace in traces.items():
  az.plot_posterior(trace, var_names=["beta0", "beta1", "beta2", "beta3", "sigma"])
  plt.suptitle(f"Posterior distributions: {prior_name}", y=1.02)
  plt.show()
````

````python
# Problem 2(e) — OLS Comparison
# Compare OLS coefficient estimates to the Bayesian posterior distributions
# using the weak_normal model from part (d)

import numpy as np
import matplotlib.pyplot as plt
import arviz as az

# --------------------------------------------------
# Fit OLS using least squares
# --------------------------------------------------
X = np.column_stack([
  np.ones(len(data)),
  x1,
  x2,
  x3
])

beta_ols, _, _, _ = np.linalg.lstsq(X, response, rcond=None)

print("OLS coefficient estimates:")
print(f"beta0 = {beta_ols[0]:.4f}")
print(f"beta1 = {beta_ols[1]:.4f}")
print(f"beta2 = {beta_ols[2]:.4f}")
print(f"beta3 = {beta_ols[3]:.4f}")

# --------------------------------------------------
# Select the Bayesian model to compare against OLS
# --------------------------------------------------
trace_e = traces["weak_normal"]

print("\nBayesian posterior summary (weak_normal prior):")
print(az.summary(trace_e, var_names=["beta0", "beta1", "beta2", "beta3", "sigma"]))

# --------------------------------------------------
# Plot 1: posterior distributions with OLS overlaid
# --------------------------------------------------
posterior_vars = ["beta0", "beta1", "beta2", "beta3"]

fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes = axes.flatten()

for i, var in enumerate(posterior_vars):
  samples = trace_e.posterior[var].values.flatten()

  axes[i].hist(samples, bins=40, density=True, alpha=0.7)
  axes[i].axvline(beta_ols[i], linestyle="--", linewidth=2)
  axes[i].set_title(f"{var}: Bayesian posterior vs OLS")
  axes[i].set_xlabel("Coefficient value")
  axes[i].set_ylabel("Density")
  axes[i].legend(["OLS estimate", "Posterior"])

plt.tight_layout()
plt.show()

# --------------------------------------------------
# Plot 2: coefficient comparison with 95% credible intervals
# --------------------------------------------------
posterior_means = []
lower_bounds = []
upper_bounds = []

for var in posterior_vars:
  samples = trace_e.posterior[var].values.flatten()
  posterior_means.append(np.mean(samples))
  lower_bounds.append(np.percentile(samples, 2.5))
  upper_bounds.append(np.percentile(samples, 97.5))

posterior_means = np.array(posterior_means)
lower_bounds = np.array(lower_bounds)
upper_bounds = np.array(upper_bounds)

errors = np.vstack([
  posterior_means - lower_bounds,
  upper_bounds - posterior_means
])

labels = ["beta0", "beta1", "beta2", "beta3"]
xpos = np.arange(len(labels))

plt.figure(figsize=(8, 5))
plt.errorbar(
  xpos,
  posterior_means,
  yerr=errors,
  fmt="o",
  capsize=5,
  label="Bayesian posterior mean ± 95% credible interval"
)
plt.scatter(
  xpos,
  beta_ols,
  marker="x",
  s=100,
  label="OLS estimate"
)
plt.xticks(xpos, labels)
plt.ylabel("Coefficient value")
plt.title("OLS vs Bayesian coefficient estimates")
plt.legend()
plt.show()
````

**g**
Ordinary least squares assumes that the linear model is correctly specified and estimates the coefficients by minimizing the sum of squared residuals.

---

Bayesian regression uses the same linear structure but treats the coefficients and residual variance as random variables with prior distributions.

---

Bayesian Inference builds on top of the priors with the likelihood function, so a more informative prior becomes harder to bend to a side, but very not informative ones take forever to gain any confidence.

---

Bayesian inference may be preferable when prior knowledge is available, when uncertainty quantification is especially important, or when the amount of data is limited and regularization is helpful.
